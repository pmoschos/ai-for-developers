## 2.2 Prompt engineering για RAG

Το prompt ενός RAG συστήματος έχει **τέσσερα κρίσιμα στοιχεία**:

1. **Persona** — ποιος είναι ο βοηθός
2. **Constraint** — να απαντά μόνο από το context
3. **Failure mode** — τι κάνει όταν δεν ξέρει
4. **Format** — πώς να γράψει την απάντηση

````python
from langchain_core.prompts import ChatPromptTemplate

RAG_TEMPLATE = """You are a helpful technical assistant for Datanous.ai.
Answer the question using ONLY the information in the context below.
Rules:
1. If the answer is not in the context, respond: "I do not have that information."
2. Do not make up numbers, features, or product names.
3. Keep answers concise and technically precise.
4. When relevant, mention the source product name.

Context:
{context}

Question: {question}

Answer:"""

rag_prompt = ChatPromptTemplate.from_template(RAG_TEMPLATE)
print("RAG prompt template created.")
print("Input variables:", rag_prompt.input_variables)
````

***Εικ. 2.1** — Το LCEL pipeline: κάθε component είναι Runnable που συνδέεται με `|`. Η παράλληλη εκτέλεση (RunnableParallel) χτίζει το context και question ταυτόχρονα.*

Το **RunnableParallel** περνάει το ίδιο input σε πολλά Runnables και επιστρέφει dictionary με τα αποτελέσματα.


### RunnablePassthrough()

```python
{
    "context": "...σχετικά documents από τον retriever...",
    "question": "What is Datanous Insight and how much does it cost?"
}
```

````python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

def format_docs(docs):
    """Join retrieved documents into a single context string."""
    return "\n\n".join(
        f"[{d.metadata.get('product', d.metadata.get('section', 'general'))}] {d.page_content}"
        for d in docs
    )

llm = ChatOpenAI(model=LLM_MODEL, temperature=0)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | rag_prompt
    | llm
    | StrOutputParser()
)

# Test the chain
answer = rag_chain.invoke("What is Datanous Insight and how much does it cost?")
print("Q: What is Datanous Insight and how much does it cost?")
print("A:", answer)
````

### Πώς διαβάζεται το πρώτο dict;

```python
{"context": retriever | format_docs, "question": RunnablePassthrough()}
```

Είναι ένα **`RunnableParallel`** — εκτελεί κάθε key παράλληλα και επιστρέφει dict.

* `"context"`: παίρνει την είσοδο, την περνά στον retriever, μετά τη μορφοποιεί
* `"question"`: απλά περνά την είσοδο όπως είναι (`RunnablePassthrough`)

Έτσι το επόμενο βήμα (`rag_prompt`) δέχεται dict με τα `{context, question}` που χρειάζεται το template.

## 2.4 Source attribution — γιατί το LLM είπε αυτό;

Σε κάθε σοβαρό RAG πρέπει να μπορείς να πεις: «Η απάντηση βασίστηκε στις πηγές A, B, C».

Αυτό απαιτεί να **κρατήσουμε** τα retrieved documents παράλληλα με την απάντηση.

````python
# Citation-aware retrieval: return answer + source documents
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

def rag_with_citations(question: str):
    """Run RAG and return the answer alongside the source documents used."""
    docs = retriever.invoke(question)
    context = format_docs(docs)

    answer = (rag_prompt | llm | StrOutputParser()).invoke(
        {"context": context, "question": question}
    )

    print(f"Q: {question}")
    print(f"A: {answer}")
    print("\nSources used:")
    for d in docs:
        src = d.metadata.get("source", "unknown")
        product = d.metadata.get("product", d.metadata.get("section", ""))
        print(f"  [{product}] {src}: {d.page_content[:80]}...")

rag_with_citations("What does Datanous Guard protect against?")
````

## 2.5 Inline citations

Πιο εξελιγμένη παραλλαγή: ζητάμε από το LLM να **παραπέμπει σε αριθμημένες πηγές** μέσα στην απάντηση.

````python
# Remind: we have declared it at the top!
RAG_TEMPLATE = """You are a helpful technical assistant for Datanous.ai.
Answer the question using ONLY the information in the context below.
Rules:
1. If the answer is not in the context, respond: "I do not have that information."
2. Do not make up numbers, features, or product names.
3. Keep answers concise and technically precise.
4. When relevant, mention the source product name.

Context:
{context}

Question: {question}

Answer:"""

rag_prompt = ChatPromptTemplate.from_template(RAG_TEMPLATE)
print("RAG prompt template created.")
print("Input variables:", rag_prompt.input_variables)
````

````python
# Score-filtered retrieval: only use documents above a relevance threshold
def rag_with_score_filter(question: str, min_score: float = 0.55):
    """Retrieve docs, filter by relevance score, then generate."""
    scored = vectorstore.similarity_search_with_relevance_scores(question, k=5)

    relevant = [(doc, score) for doc, score in scored if score >= min_score]
    print(f"Q: {question}")
    print(f"Retrieved {len(scored)} docs, {len(relevant)} above threshold ({min_score}):")
    for doc, score in relevant:
        print(f"  [{score:.3f}] {doc.page_content[:80]}")

    if not relevant:
        print("No sufficiently relevant documents found.")
        return

    context = "\n\n".join(doc.page_content for doc, _ in relevant)
    answer = (rag_prompt | llm | StrOutputParser()).invoke(
        {"context": context, "question": question}
    )
    print(f"\nA: {answer}")

rag_with_score_filter("What is the price of Datanous Insight Professional?")
````

## 2.6 Structured output — JSON αντί για κείμενο

Συχνά θέλουμε η απάντηση να είναι structured (πχ για downstream API).
Με το `with_structured_output()` του LangChain, παίρνουμε type-safe Pydantic objects.

````python
from pydantic import BaseModel, Field
from typing import Optional

class DatanousProductAnswer(BaseModel):
    """Structured answer about a Datanous.ai product."""
    product_name: str = Field(description="Name of the Datanous.ai product")
    pricing:      Optional[str] = Field(None, description="Pricing information if mentioned")
    uptime_sla:   Optional[str] = Field(None, description="Uptime SLA if mentioned")
    key_features: list[str]     = Field(default_factory=list, description="Main features listed")
    summary:      str           = Field(description="One or two sentence summary")

structured_llm = llm.with_structured_output(DatanousProductAnswer)

STRUCT_PROMPT = ChatPromptTemplate.from_template(
    """Given the context below, extract structured information about the Datanous.ai product.
If a field is not mentioned, leave it empty.

Context:
{context}

Question: {question}"""
)

struct_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | STRUCT_PROMPT
    | structured_llm
)

result = struct_chain.invoke("Tell me about Datanous Insight pricing and features.")
print("Structured answer:")
print(f"  Product      : {result.product_name}")
print(f"  Pricing      : {result.pricing}")
print(f"  Uptime SLA   : {result.uptime_sla}")
print(f"  Key features : {result.key_features}")
print(f"  Summary      : {result.summary}")
````

## 2.7 Streaming — απόκριση token-token

Για UX, οι χρήστες θέλουν να βλέπουν την απάντηση να εμφανίζεται **σταδιακά**, όχι να περιμένουν 5 δευτερόλεπτα.

Το LCEL υποστηρίζει streaming out-of-the-box — αρκεί να χρησιμοποιείς `.stream()` αντί για `.invoke()`.

````python
# Streaming: print tokens as they arrive instead of waiting for the full response
print("Streaming answer:\n")
for chunk in rag_chain.stream("What are the four checks that Datanous Guard performs?"):
    print(chunk, end="", flush=True)
print()
````

## 2.8 Batch — πολλές ερωτήσεις παράλληλα

Όταν θες να τρέξεις πολλές ερωτήσεις (πχ για evaluation), `.batch()` τις τρέχει **παράλληλα**.

````python
# Batch: run multiple questions in parallel (max_concurrency controls parallelism)
questions = [
    "What is the price of the Datanous Insight Professional plan?",
    "What embedding dimensions does Datanous Embed support?",
    "Which industries does Datanous.ai serve?",
]

answers = rag_chain.batch(questions, config={"max_concurrency": 3})
print("Batch answers:")
for q, a in zip(questions, answers):
    print(f"\nQ: {q}")
    print(f"A: {a}")
````

## 2.9 Conversational RAG — μνήμη συνομιλίας

Σε ένα chat, η νέα ερώτηση συχνά **εξαρτάται** από την προηγούμενη.

Παράδειγμα:

> User: What is Datanous Insight?  
> Bot: Datanous Insight is a RAG-powered knowledge management platform that indexes enterprise documents and answers queries with source citations.  
> User: **And how much does it cost for 100,000 pages?**

Η δεύτερη ερώτηση έχει νόημα **μόνο** αν ξέρουμε ότι το `it` αναφέρεται στο **Datanous Insight**.

Ένας naive retriever θα αναζητούσε κάτι σαν:

> And how much does it cost for 100,000 pages?

Το πρόβλημα είναι ότι η ερώτηση περιέχει αντωνυμία (`it`) και δεν είναι πλήρως αυτόνομη. Αν περάσει έτσι στον retriever, μπορεί να μη βρει το σωστό pricing document.

**Λύση:** δύο βήματα.

1. **Reformulate**: ένα LLM call ξαναγράφει την ερώτηση ως standalone.
2. **RAG**: ο retriever ψάχνει τη standalone version.

````python
from langchain_core.prompts import MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

REFORMULATE_TEMPLATE = ChatPromptTemplate.from_messages([
    ("system",
     "Given the chat history and the latest user question, rewrite the question "
     "so it is fully self-contained (standalone) — answerable WITHOUT the history. "
     "Do NOT answer the question. Return only the rewritten question. "
     "If the question is already standalone, return it unchanged."),
    MessagesPlaceholder("chat_history"),
    ("human", "{question}"),
])

reformulate_chain = REFORMULATE_TEMPLATE | llm | StrOutputParser()

# Demo: simulate a follow-up question about Datanous Insight
history = [
    HumanMessage(content="What is Datanous Insight?"),
    AIMessage(content="Datanous Insight is a RAG-powered knowledge management platform "
                      "that indexes enterprise documents and answers queries with source citations."),
]

follow_up = "And how much does it cost for 100,000 pages?"
standalone = reformulate_chain.invoke({"chat_history": history, "question": follow_up})
print(f"Original follow-up  : {follow_up}")
print(f"Standalone rewrite  : {standalone}")
````

Τώρα συνδυάζουμε reformulate + retrieve + answer σε ένα chain.

````python
QA_WITH_HISTORY = ChatPromptTemplate.from_messages([
    ("system",
     "You are the technical assistant for Datanous.ai. "
     "Answer questions using ONLY the context below. "
     "If you do not know, say so.\n\nContext:\n{context}"),
    MessagesPlaceholder("chat_history"),
    ("human", "{question}"),
])

def conversational_rag(question: str, chat_history: list) -> str:
    """Full conversational RAG: reformulate if needed, retrieve, generate."""
    # Step 1: Reformulate if there is history
    standalone_q = (
        reformulate_chain.invoke({"chat_history": chat_history, "question": question})
        if chat_history else question
    )

    # Step 2: Retrieve using the standalone question
    docs    = retriever.invoke(standalone_q)
    context = format_docs(docs)

    # Step 3: Generate with full history context
    chain  = QA_WITH_HISTORY | llm | StrOutputParser()
    return chain.invoke({
        "context":      context,
        "chat_history": chat_history,
        "question":     question,
    })

# Demo: multi-turn conversation about Datanous.ai products
chat_history = []

turns = [
    "What is Datanous Insight?",
    "What does the Professional plan include?",
    "Is there an Enterprise option with better uptime?",
    "What about data privacy — is it GDPR compliant?",
]

for q in turns:
    answer = conversational_rag(q, chat_history)
    print(f"User : {q}")
    print(f"Bot  : {answer}\n")
    chat_history.append(HumanMessage(content=q))
    chat_history.append(AIMessage(content=answer))
````

````python
QA_NO_HISTORY = ChatPromptTemplate.from_messages([
    ("system",
     "You are the technical assistant for Datanous.ai. "
     "Answer questions using ONLY the context below. "
     "If you do not know, say so.\n\nContext:\n{context}"),
    ("human", "{question}"),
])

def rag_without_history(question: str) -> str:
    """Simple RAG: retrieve directly from the original question, then generate."""
    
    # Step 1: Retrieve directly using the user's original question
    docs = retriever.invoke(question)
    context = format_docs(docs)

    # Step 2: Generate answer without chat history
    chain = QA_NO_HISTORY | llm | StrOutputParser()

    return chain.invoke({
        "context": context,
        "question": question,
    })
````

````python
turns = [
    "What is Datanous Insight?",
    "What does the Professional plan include?",
    "Is there an Enterprise option with better uptime?",
    "What about data privacy — is it GDPR compliant?",
]

for q in turns:
    answer = rag_without_history(q)
    print(f"User : {q}")
    print(f"Bot  : {answer}\n")
````

## 2.10 Pitfalls στο RAG generation

Αυτά τα προβλήματα εμφανίζονται **ακόμα και με τέλειο retrieval**. Φύλαξε αυτή τη λίστα:

| Pitfall | Πώς εκδηλώνεται | Πώς το αντιμετωπίζεις |
|---|---|---|
| **Hallucination** | Το LLM προσθέτει «νούμερα» που δεν υπάρχουν | Auster prompt: «μόνο από context», temperature=0 |
| **Lost in the middle** | Αν δώσεις 10 docs, αγνοεί τα μεσαία | Re-rank ή fewer docs (notebook 06) |
| **Conflicting context** | 2 docs λένε διαφορετικά νούμερα | Πρόσθεσε στο prompt: «αν συγκρούονται, ζήτα διευκρίνιση» |
| **Stale data** | Index έχει παλιά τιμή → λάθος απάντηση | Re-index regularly + metadata `last_updated` |
| **Sycophancy** | Συμφωνεί με κάθε ισχυρισμό του χρήστη | Πρόσθεσε «μην συμφωνείς εκτός αν επιβεβαιώνεται από context» |
| **Context dilution** | Πολύ μεγάλο context → χάνει εστίαση | Compression / summarization (notebook 05) |
